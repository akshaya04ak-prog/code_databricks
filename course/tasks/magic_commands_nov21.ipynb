```python
import requests

# Pull the CSV data from the public URL into a variable
response = requests.get("https://public.tableau.com/app/sample-data/mobile_os_usage.csv")

# Write the data into the created volume
path = "/Volumes/workspace/default/usage_metric/mobile_os_usage.csv"
dbutils.fs.put(path, response.text, overwrite=True)

print(f"Data written to {path}")
# Show only the first few lines of the file
print("\n".join(dbutils.fs.head(path).splitlines()[:2]))
```
